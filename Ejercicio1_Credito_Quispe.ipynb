## Ejercicio 1: Evaluación de solicitud de crédito bancario

### Ficha PEAS
| Componente | Descripción |
| :--- | :--- |
| **Percepción (S)** | `ingreso_mensual` (soles), `monto_solicitado` (soles), `tiene_historial_moroso` (booleano) |
| **Acciones (A)** | Aprobar solicitud, Aprobar con condiciones, Rechazar solicitud |
| **Entorno (E)** | Sistema de evaluación crediticia y base de datos financiera del banco |
| **Objetivo** | Maximizar el retorno financiero por préstamos reduciendo la tasa de morosidad |
| **Medida de desempeño (P)** | Porcentaje de créditos recuperados a tiempo y reducción de pérdidas por impago |

### Justificación de las reglas
Las reglas se basan en la capacidad de pago y el riesgo crediticio: si el solicitante registra historial moroso, representa un riesgo crítico que exige rechazo si el préstamo supera su ingreso, o aprobación condicionada únicamente con montos muy bajos. Un endeudamiento moderado sin morosidad es ideal para la aprobación directa, manteniendo la solvencia del cliente y la rentabilidad del banco.

### Código

```python
def agente_credito(ingreso_mensual, monto_solicitado, tiene_historial_moroso):
    if ingreso_mensual <= 0:
        return "rechazar", "Ingreso mensual inválido o nulo."

    relacion = monto_solicitado / ingreso_mensual

    if tiene_historial_moroso:
        if relacion <= 0.2:
            return (
                "aprobar con condiciones",
                f"Historial moroso registrado, pero la relación monto/ingreso es baja ({relacion:.2f}). Requiere aval.",
            )
        else:
            return (
                "rechazar",
                f"Historial moroso registrado y la relación monto/ingreso es elevada ({relacion:.2f}). Risk score alto.",
            )
    else:
        if relacion <= 0.5:
            return (
                "aprobar",
                f"Sin morosidad y relación monto/ingreso saludable ({relacion:.2f}).",
            )
        elif relacion <= 1.0:
            return (
                "aprobar con condiciones",
                f"Sin morosidad, pero la relación monto/ingreso es moderada/alta ({relacion:.2f}). Requiere garantía.",
            )
        else:
            return (
                "rechazar",
                f"Monto solicitado excede la capacidad de pago (relación {relacion:.2f}).",
            )
```

```python
casos_prueba = [
    (5000, 1500, False),
    (3000, 2400, False),
    (2000, 3000, False),
    (4000, 500, True),
    (4000, 2000, True),
]

for i, (ingreso, monto, moroso) in enumerate(casos_prueba, 1):
    accion, motivo = agente_credito(ingreso, monto, moroso)
    print(f"Caso {i} -> Decision: {accion.upper()} | Motivo: {motivo}")
```

### Visualización de decisiones

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
n = 150
ingresos = np.random.uniform(1000, 10000, n)
montos = np.random.uniform(500, 12000, n)
morosos = np.random.choice([True, False], size=n, p=[0.3, 0.7])

acciones = [
    agente_credito(ing, mon, mor)[0]
    for ing, mon, mor in zip(ingresos, montos, morosos)
]
colores = {
    "aprobar": "green",
    "aprobar con condiciones": "orange",
    "rechazar": "red",
}

plt.figure(figsize=(9, 5))
for acc, col in colores.items():
    idx = [i for i, a in enumerate(acciones) if a == acc]
    plt.scatter(
        ingresos[idx],
        montos[idx] / ingresos[idx],
        c=col,
        label=acc.capitalize(),
        alpha=0.7,
    )

plt.axhline(0.5, color="gray", linestyle="--", label="Límite 0.5")
plt.axhline(1.0, color="black", linestyle="--", label="Límite 1.0")
plt.title("Evaluación de Créditos: Relación Monto/Ingreso vs Ingreso")
plt.xlabel("Ingreso Mensual (Soles)")
plt.ylabel("Relación (Monto / Ingreso)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
